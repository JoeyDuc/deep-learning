```python
import numpy as np
import matplotlib.pyplot as plt
errors = np.linspace(-5, 5, 1000)

# Hai hàm loss
squared_loss = errors ** 2
absolute_loss = np.abs(errors)

# Visualization
plt.figure(figsize=(9, 5))

plt.plot(errors, squared_loss, label = 'Squared Loss')
plt.plot(errors, absolute_loss, label = 'Absolute Loss')

plt.xlabel('Prediction Error')
plt.ylabel('Loss')
plt.title('Squared Loss vs Absolute Loss')
plt.axhline(0, linewidth = 0.8)
plt.axvline(0, linewidth = 0.8)

plt.legend()
plt.grid(alpha = 0.3)
plt.show()
```

```python

def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

# f(x) = x^2
f = lambda x: x**2

for x in [-2, -1, 0, 1, 2]:
    approx = numerical_derivative(f, x)
    exact = 2 * x

    print(
        f"x={x}, "
        f"Numerical={approx:.4f}, "
        f"Exact={exact:.4f}"
    )
```
