```python
import numpy as np
import matplotlib.pyplot as plt

def get_column(data, index):
    result = [row[index] for row in data]
    return result

data = np.genfromtxt('data/data.csv', delimiter=',').tolist()

x_data = get_column(data, 0)
y_data = get_column(data, 1)
N = len(y_data)

print(f'areas: {x_data}')
print(f'prices: {y_data}')
print(f'data_size: {N}')
```

```python
def predict(x, w, b):
    return w * x + b

def compute_loss(y_hat, y, delta = 5):
    if abs(y_hat - y) < delta:
        loss = (y_hat - y) ** 2
    else:
        loss = delta * abs(y_hat - y) - 0.5 * delta * delta
    return loss

# compute gradient: square error
def gradient_sr(y_hat, y, x):
    dw = 2 * x * (y_hat - y)
    db = 2 * (y_hat - y)
    return dw, db

# compute gradient: absolute error
def gradient_ab(y_hat, y, x, delta):
    dw = delta * (y_hat - y) * x / abs(y_hat - y)
    db = delta * (y_hat - y) / abs(y_hat - y)
    return dw, db

def gradient(y_hat, y, x, delta = 5):
    if abs(y_hat - y) < delta:
        dw, db = gradient_sr(y_hat, y, x)
    else:
        dw, db = gradient_ab(y_hat, y, x, delta)
    return dw, db


def update_weights(w, b, dw, db, lr):
    new_w = w - lr * dw
    new_b = b - lr * db
    return new_w, new_b

```

```python
b = 0.04
w = -0.34
lr = 0.01

epoch_max = 10
losses = []

for epoch in range(epoch_max):
    for i in range(N):
        x = x_data[i]
        y = y_data[i]

        y_hat = predict(x, w, b)

        loss = compute_loss(y_hat, y)

        losses.append(loss)

        dw, db = gradient(y_hat, y, x)
        w, b = update_weights(w, b, dw, db, lr)
print(f'w = {w}, b = {b}')
```

```python
plt.plot(losses)
plt.xlabel('iteration')
plt.ylabel('losses')
plt.show()
```
