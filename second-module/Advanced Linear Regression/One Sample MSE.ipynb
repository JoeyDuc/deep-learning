```python
import numpy as np
import matplotlib.pyplot as plt

def get_column(data, index):
    result = [row[index] for row in data]
    return result

data = np.genfromtxt('data/data.csv', delimiter=',').tolist()

areas = get_column(data, 0)
prices = get_column(data, 1)

N = len(areas)

print(f'areas: {areas}')
print(f'prices : {prices}')
print(f'data_size: {N}')
```

```python
plt.scatter(areas, prices)
plt.xlabel('Area (x 100$m^2$)')
plt.ylabel('Price (Tael)')
plt.xlim(3, 7)
plt.ylim(4, 10)
plt.show()
```

```python
def predict(x, w, b):
    return x * w + b

def gradient(y_hat, y, x):
    dw = 2 * x * (y_hat - y)
    db = 2 * (y_hat - y)
    return dw, db

def update_weight(w, b, dw, db, lr):
    new_w = w - lr * dw
    new_b = b - lr * db
    return new_w, new_b
```

```python
b = 0.04
w = -0.34
lr = 0.02

epoch_max = 10
losses = []
for epoch in range(epoch_max):
    for i in range(N):
        x = areas[i]
        y = prices[i]

        y_hat = predict(x, w, b)

        loss = (y_hat - y) ** 2
        print(loss)
        losses.append(loss)

        dw, db = gradient(y_hat, y, x)

        w, b = update_weight(w, b, dw, db, lr)
```

```python
print(f'w = {w}, b = {b}')
```
