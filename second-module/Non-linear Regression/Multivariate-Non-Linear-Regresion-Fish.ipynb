```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('data/Fish.csv')
df
```

```python
encoded_species = pd.get_dummies(df.Species)
encoded_species.head(5)
```

```python
new_df = pd.concat([df, encoded_species], axis = 1)
new_df
```

```python
X = new_df[['VerticalLen',	'DiagonalLen',	'CrossLen',	'Height', 	'Width', 'Bream', 'Parkki', 'Perch', 'Pike', 'Roach', 'Smelt', 'Whitefish']]
y = new_df['Weight']
```

```python
X = X.values
y = y.values
```

```python
X
```

```python
y
```

```python
X_data = np.hstack((np.ones((X.shape[0], 1)), X))
X_data
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_data, y, test_size = 0.2, random_state = 32)
print('X_train shape: ', X_train.shape)
print(f'X_test shape: {X_test.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'y_test shape : {y_test.shape}')
```

```python
def r2score(y_pred, y):
    rss = np.sum((y_pred - y) ** 2)
    tss = np.sum((y - y.mean()) ** 2)
    r2 = 1 - (rss / tss)
    return r2
```

```python
class LinearRegression:
    def __init__(self, X_data, y_target, learning_rate = 0.00001, num_epochs = 10000):
        self.X_data = X_data
        self.y_target = y_target
        self.learning_rate = learning_rate
        self.num_epochs = num_epochs
        self.num_samples = self.X_data.shape[0]

        # Initial Coefficient
        self.theta = np.random.randn(self.X_data.shape[1]) # [b w]
        self.losses = []
    def compute_loss(self, y_pred, y_target):
        loss = (y_pred - y_target) ** 2
        loss = np.mean(loss)
        return loss
    def predict(self, X_data):
        y_pred = X_data.dot(self.theta)
        return y_pred
    def fit(self):
        for epoch in range(self.num_epochs):
            # predict
            y_pred = self.predict(self.X_data)

            # compute loss
            loss = self.compute_loss(y_pred, self.y_target)
            self.losses.append(loss)

            # compute gradient
            k = 2 * (y_pred - self.y_target)
            gradients = self.X_data.T.dot(k) / self.num_samples

            # update weights
            self.theta = self.theta - self.learning_rate * gradients

            print(f'Epoch: {epoch} - Loss : {loss}')
        return {
            'loss': sum(self.losses) / len(self.losses),
            'weight': self.theta
        }
```

```python
linear_model = LinearRegression(X_train, y_train, learning_rate = 0.0001, num_epochs = 100)
linear_model.fit()
```

```python
preds = linear_model.predict(X_train)
r2score(preds, y_train)
```

```python
preds =  linear_model.predict(X_test)
r2score(preds, y_test)
```

```python
# Simple Approach
# from (a + b) ^ 2 = a ^ 2 + b ^ 2 + a + b + 1

def create_polynomial_features(X, degree = 2):
    X_mem = []
    for X_sub in X.T:
        X_new = X_sub
        for d in range(2, degree + 1):
            X_new = np.c_[X_new, np.power(X_sub, d)]
        X_mem.extend(X_new.T)
    return np.c_[X_mem].T
```

```python
X_new = np.array([
    [1, 2],
    [3, 4]
])
X_new
```

```python
create_polynomial_features(X_new, degree = 2)
```

```python
create_polynomial_features(X_new, degree = 3)
```

```python
X_poly = create_polynomial_features(X, degree = 2)
X_poly
```

```python
X_data = np.hstack((np.ones((X_poly.shape[0], 1)), X_poly))
X_data
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_data, y, test_size = 0.2, random_state = 32)
print(f'X_train shape : {X_train.shape}')
print(f'X_test shape: {X_test.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'y_test shape: {y_test.shape}')
```

```python
poly_model = LinearRegression(X_train, y_train, learning_rate = 0.0000001, num_epochs = 100)
poly_model.fit()
```

```python
preds = poly_model.predict(X_train)
r2score(preds, y_train)
```

```python
preds = poly_model.predict(X_test)
r2score(preds, y_test)
```
