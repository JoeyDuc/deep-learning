```python
def f_function(a, x):
    y = a * x
    return y
def loss(y, f):
    return (f - y) ** 2
def compute_dg_da(x, f, y):
    return 2 * x * (f - y)
```

```python
x = 4
y = 6
a = -6
lr = 0.01

print(f'f(x) = {f_function(a, x)}')
print(f'loss = {loss(y, f_function(a, x))}')
```

```python
f = f_function(a, x)
dg_da = compute_dg_da(x, f, y)
a = a - lr * dg_da

print(f'loss = {loss(y, f_function(a, x))}')
```

```python
import matplotlib.pyplot as plt
import numpy as np

x_values = np.arange(-10, 10, 0.1)
y_values = x_values * a

plt.plot(x_values, y_values)
plt.scatter([4], [6])
plt.grid()
plt.show()
```

```python
x = 4
y = 6

a_values = np.arange(-10, 10, 0.1)
g_value = (a_values * x - y) ** 2
dg_da_values = 2 * x * (a_values * x - y)
```

```python
plt.plot(a_values, g_value)
plt.plot(a_values, dg_da_values)
plt.grid()
plt.show()
```
