```python
import random

# predict function
def predict_func(x, a, b):
    return a * x + b

# parameters
num_steps = 100
lr = 0.01

# given data
data = [
    [-2, -1],
    [5, 13]
]

# 1 set a, b, randomly
a = random.random() * 10 - 5
b = random.random() * 10 - 5
 # Sequence
for i in range(num_steps):
    for sample in data:
        x_value, y_value = sample

        # compute_predicted_y
        predicted_y = predict_func(x_value, a, b)

        # compute g
        g_value = (predicted_y - y_value) ** 2

        # compute partial gradients for a and b
        dg_da = 2 * x_value * (predicted_y - y_value)
        dg_db = 2 * (predicted_y - y_value)

        # update
        a = a - lr * dg_da
        b = b - lr * dg_db
```

```python
x1, y1 = -2, -1
x2, y2 = 5, 13
# Parallel

for i in range(num_steps):
    predicted_y1 = predict_func(x1, a, b)
    predicted_y2 = predict_func(x2, a, b)

    g_value_1 = (predicted_y1 - y1) ** 2
    g_value_2 = (predicted_y2 - y2) ** 2

    dg_da = 2 * x1 * (predicted_y1 - y1) + 2 * x2 * (predicted_y2 - y2)
    dg_db = 2 * (predicted_y1 - y1) + 2 * (predicted_y2 - y2)

    a = a - lr * dg_da / 2
    b = b - lr * dg_db / 2
```
