```python
import matplotlib.pyplot as plt
import numpy as np

areas = [6.7, 4.6, 3.5, 5.5]
prices = [9.1, 5.9, 4.6, 6.7]
w = 5
b = 0.04

x_values = np.arange(3, 7, 0.1)
y_values = x_values * w + b

plt.scatter(areas, prices)
plt.plot(x_values, y_values, 'g')
```

```python
N = 4
num_epochs = 10
lr = 0.01
w = -0.34
b = 0.04

def predict(x, w, b):
    return w * x - b

def compute_loss(y_hat, y):
    return (y_hat - y) ** 2

def compute_gradient(x, y_hat, y):
    dl_dw = 2 * x * (y_hat - y)
    dl_db = 2 * (y_hat - y)
    return dl_dw, dl_db # return tupple

def update(w, b, dw, db, lr):
    new_w = w - lr * dw
    new_b = b - lr * db
    return new_w, new_b

losses = []

for _ in range(num_epochs):
    for i in range(N):
        x = areas[i]
        y = prices[i]
        print(f'x = {x}, y = {y}')
        # predict
        y_hat = predict(x, w, b)
        print('y_hat = {y_hat}')

        # compute Loss
        loss = compute_loss(y_hat, y)
        losses.append(loss)

        # compute_gradient
        dl_dw, dl_db = compute_gradient(x, y_hat, y)
        print(f'dl_dw = {dl_dw}, dl_db = {dl_db}')
        # update
        w, b = update(w, b, dl_dw, dl_db, lr)
        print(f'new_w = {w}, new_b = {b}')
```
