```python
lr = 0.01
w = -0.34
b = 0.04

def predict(x, w, b):
    return w * x - b

def compute_loss(y_hat, y):
    return (y_hat - y) ** 2

def compute_gradient(x, y_hat, y):
    dl_dw = 2 * x * (y_hat - y)
    dl_db = 2 * (y_hat - y)
    return dl_dw, dl_db # return tupple

def update(w, b, dw, db, lr):
    new_w = w - lr * dw
    new_b = b - lr * db
    return new_w, new_b
```

```python
areas = [6.7, 4.6, 3.5, 5.5]
prices = [9.1, 5.9, 4.6, 6.7]

epoch = 100
m = 2
N = 4

losses = []

for _ in range(epoch):
    # for each epoch
    for i in range(0, N, m):
        # for each mini-batch
        acc_loss = 0
        acc_dl_dw = 0
        acc_dl_db = 0
        for j in range(m):
            # get a sample
            x = areas[i + j]
            y = prices[i + j]

            # predict
            y_hat = predict(x, w, b)

            # compute_loss
            loss = compute_loss(y_hat, y)
            acc_loss += loss

            # compute gradient
            dl_dw, dl_db = compute_gradient(x, y_hat, y)

            acc_dl_dw += dl_dw
            acc_dl_db += dl_db
        # compute average
        avg_loss = acc_loss / m

        losses.append(avg_loss)

        avg_dl_dw = acc_dl_dw / m
        avg_dl_db = acc_dl_db / m
        # update
        w, b = update(w, b, avg_dl_dw, avg_dl_db, lr)

        break
```

```python
import matplotlib.pyplot as plt
plt.plot(losses)
```
