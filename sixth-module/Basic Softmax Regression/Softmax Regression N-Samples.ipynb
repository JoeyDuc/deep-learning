```python
import numpy as np
import numpy as np
X = np.array([
    [0.5],
    [1.0],
    [1.5],
    [2.0],
    [2.5],
    [3.0],
    [3.5],
    [4.0]
])
Y = np.array([0, 0, 1, 0, 0, 1, 1, 1])
```

```python
def convert_one_hot(y, k):
    one_hot = np.zeros((len(y), k))
    one_hot[np.arange(len(y)), y] = 1
    return one_hot

n_classes = 2
Y_onehot = convert_one_hot(Y, n_classes)
Y_onehot
```

```python
X = np.hstack([np.ones((len(X), 1)), X])
X.shape
```

```python
theta = np.array([[0.1, 0.2], [0.3, 0.4]])
```

```python

def softmax_function(z):
    return np.exp(z) / np.sum(np.exp(z), axis = 1, keepdims=True)

z = X.dot(theta)
print(z)
```

```python
print(softmax_function(z))
```

```python
def predict(x, theta):
    z = X.dot(theta)
    y_hat = softmax_function(z)
    return y_hat

y_hat = predict(X, theta)
print(y_hat)
```

```python
def compute_loss(y_hat, y):
    loss = -np.log(np.sum(y_hat * y, axis = 1))
    return loss

compute_loss(y_hat, Y_onehot)
```

```python
def compute_gradient(y_hat, y, x):
    gradient = x.T.dot(y_hat - y)
    return gradient

print(compute_gradient(y_hat, Y_onehot, X))
```

```python
def update_weight(theta, gradient, lr):
    theta = theta - lr * gradient
    return theta
```

```python
learning_rate = 0.01
n_epochs = 100
losses = []
for epoch in range(n_epochs):
    y_hat = predict(X, theta)

    loss = compute_loss(y_hat, Y_onehot)
    losses.append(sum(loss))

    gradient = compute_gradient(y_hat, Y_onehot, X)

    theta = update_weight(theta, gradient, learning_rate)
```

```python
import matplotlib.pyplot as plt

plt.plot(losses)
```
