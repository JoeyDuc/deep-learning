```python
# Broadcasting
tensor1 = torch.tensor([
    [1, 2],
    [3, 4]
])
tensor2 = torch.tensor([1])
print(tensor1 + tensor2)
```

```python
x = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = 5

squared_diff = (x - y) ** 2
print(f'x:\n {x}')
print(f'y:\n {y}')
print(f'squared_diff:\n {squared_diff}')
```

```python
x = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = torch.tensor([5, 5, 5, 5], dtype=torch.float32)

loss_fn = torch.nn.MSELoss()
mse = loss_fn(x, y)
print(f'x:\n {x}')
print(f'y:\n {y}')
print(f'mse:\n {mse}')
```

```python
x = torch.tensor([[1, 2],
                  [3, 4]])
y = torch.tensor([[3, 4],
                  [5, 6]])

tensor1 = torch.cat((x, y), dim = 0)
print(f'Tensor1:\n {tensor1}')

tensor2 = torch.cat((x, y), dim = 1)
print(f'Tensor2:\n {tensor2}')
```

```python
data = torch.randint(low = 0, high=9, size = (3, 2))
print(f'data:\n {data}')

argmax_dim0 = torch.argmax(data, dim = 0)
print(f'argmax0: \n {argmax_dim0}')

argmax_dim1 = torch.argmax(data, dim = 1)
print(f'argmax1:\n {argmax_dim1}')
```

```python
# Create a tensor to compute gradients
x = torch.tensor(2.0, requires_grad = True)
print(x)

# operation
y = x ** 2
print(y)
z = 3 * y + 2
print(f'z: {z}')

# Backpropagate to compute gradients
z.backward()
print(f'Gradient of z w.r.t. x: {x.grad}')
```

```python
x = torch.tensor([6.7])
print(x)

w = -0.34
b = 0.04
y_hat = x * w + b
print(y_hat)
```

```python
import torch.nn as nn 

linear_model = nn.Linear(1, 1)

print(linear_model.bias)
print(linear_model.weight.shape)

linear_model.bias.data = torch.tensor([1], dtype=torch.float32)
linear_model.weight.data = torch.tensor([[0.3]], dtype = torch.float32)
print("__________________________")
print(linear_model.bias)
print(linear_model.weight)
```

```python
y_hat = linear_model(x)
print(y_hat)
```
