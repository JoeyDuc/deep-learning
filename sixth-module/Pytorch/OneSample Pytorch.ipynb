```python
import numpy as np
import torch
import torch.nn as nn 

data = np.genfromtxt('data/data.csv', delimiter=',')
x_data = torch.from_numpy(data[:, 0:1]).float()
y_data = torch.from_numpy(data[:, 1]).float()
print(x_data)
print(y_data)
```

```python
linear = nn.Linear(1, 1)

linear.weight.data = torch.Tensor([[-0.34]])
linear.bias.data = torch.Tensor([0.04])

print(linear.weight)
print(linear.bias)
```

```python
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(linear.parameters(), lr = 0.01)
epochs = 1
```

```python
# training
for epoch in range(epochs):
    for i in range(len(x_data)):
        x = x_data[i]
        y = y_data[i]

        # y_hat
        y_hat = linear(x)
        print(f'y_hat: {y_hat}')

        # loss
        loss = loss_fn(y_hat, y)
        print(f'loss: {loss}')

        # compute gradient
        optimizer.zero_grad()
        loss.backward()
        print(f'(db, dw): ({linear.bias.grad.data}, {linear.weight.grad.data})')

        # update
        optimizer.step()
        print(f'(b, w): ({linear.bias.data}, {linear.weight.data})\n\n')
```
