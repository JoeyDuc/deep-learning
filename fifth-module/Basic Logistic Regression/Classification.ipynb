```python
X = df[['Width', 'Length']]
X.shape
```

```python
# Standarize
mean = X.mean(axis = 0)
std = X.std(axis = 0)
X = (X - mean) / std
print(X)
```

```python
y = df['Result']
y.shape
```

```python
X = np.hstack([np.ones((len(X), 1)), X])
print(X.shape)
```

```python
n_features = X.shape[-1]
theta = np.random.rand(n_features)
print(theta)
print(theta.reshape(-1, 1).shape)
```

```python
def logistic_function(z):
    return 1 / (1 + np.exp(-z))

def predict(x, theta):
    z = np.dot(x, theta)
    y_hat = logistic_function(z)
    return y_hat
def compute_loss(y_hat, y):
    loss = sum(-1 * (y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))) / len(y)
    return loss

def compute_gradient(x, y_hat, y):
    gradient = np.dot(x.T, (y_hat - y)) / len(y)
    return gradient

def update_weight(theta, gradient, lr):
    theta = theta - lr * gradient
    return theta
```

```python
epoch_max = 10
lr = 0.1
losses = []
for epoch in range(epoch_max):
    # for each epoch - n samples
    y_hat = predict(X, theta)

    loss = compute_loss(y_hat, y)

    losses.append(loss)

    gradient = compute_gradient(X, y_hat, y)

    theta = update_weight(theta, gradient, lr)
print(losses)
```

```python
print(theta)
```

```python
y_preds = predict(X, theta)
y_preds[:5]
```

```python
thresholds = [0.1, 0.45, 0.5, 0.55, 0.65, 0.7, 0.75, 0.8]
accs = []
for threshold in thresholds:
    y_pre_th = []
    for y_pred in y_preds:
        if y_pred >= threshold:
            y_pre_th.append(1)
        else:
            y_pre_th.append(0)
    y_pred_th = np.asarray(y_pre_th)
    acc = sum(y_pre_th == y) / len(y)
    accs.append(acc)
accs
```
