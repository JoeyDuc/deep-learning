```python
import numpy as np

X = np.array([1, 0.5])
y = np.array([0])
print(X)
print(y)
```

```python
theta = np.array([0.1, 0.1])
theta
```

```python
z = np.dot(X, theta)
print(z)
```

```python
def logistic_function(x):
    return 1 / (1 + np.exp(-x))

y_hat = logistic_function(z)
print(y_hat)
```

```python
def compute_loss(y_hat, y):
    loss = -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))
    return loss
loss = compute_loss(y_hat, y)
print(loss)
```

```python
def compute_derivative(y_hat, y, x):
    return x * (y_hat - y)
derivative = compute_derivative(y_hat, y, X)
print(derivative)
```

```python
def update_weights(theta, derivative, lr):
    theta = theta - lr * derivative
    return theta
theta = update_weights(theta, derivative, 0.1)
print(theta)
```
