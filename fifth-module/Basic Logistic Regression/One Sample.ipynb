```python
import numpy as np

X = np.array([
    [1.0, 0.5],
    [2.0, 1.0],
    [3.0, 1.5],
    [2.0, 2.0],
    [1.0, 2.5],
    [2.0, 3.0],
    [1.0, 3.5],
    [2.0, 4.0]
])
Y = np.array([0, 0, 1, 0, 0, 1, 1, 1])
```

```python
n_samples = X.shape[0]
print(n_samples)
```

```python
X = np.hstack([np.ones((n_samples, 1)), X])
print(X)
```

```python
n_features = X.shape[1]
print(n_features)
```

```python
theta = np.random.rand(n_features)
print(theta)
```

```python
def logistic_function(x):
    return 1 / (1 + np.exp(-x))

def predict(x, theta):
    z = np.dot(x, theta)
    y_hat = logistic_function(z)
    return z, y_hat

def compute_loss(y_hat, y):
    loss = -1 * (y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))
    return loss

def compute_gradient(x, y_hat, y):
    gradient = x * (y_hat - y)
    return gradient

def update_weight(gradient, theta, lr):
    theta = theta - lr * gradient
    return theta
```

```python
n_epochs = 10
lr = 0.1
losses = []
for epoch in range(n_epochs):
    losses_epoch = []
    for i in range(n_samples):
        x = X[i, :]
        y = Y[i]

        z, y_hat = predict(x, theta)
        loss = compute_loss(y_hat, y)

        losses_epoch.append(loss)

        gradient = compute_gradient(x, y_hat, y)

        theta = update_weight(gradient, theta, lr)
    losses.append(sum(losses_epoch) / len(losses_epoch))

losses
```
