```python
import numpy as np
X = np.array([[1.0, 0.5],
             [2.0, 1.0],
             [3.0, 1.5],
             [2.0, 2.0],
             [1.0, 2.5],
             [2.0, 3.0],
             [1.0, 3.5],
             [2.0, 4.0]])

Y = np.array([0, 0, 1, 0, 0, 1, 1, 1])
```

```python
print(X)
```

```python
n_samples = X.shape[0]
n_samples
```

```python
X = np.hstack([np.ones((n_samples, 1)), X])
print(X)
```

```python
n_features = X.shape[1]
print(n_features)
```

```python
theta = np.random.rand(n_features)
print(theta)
```

```python
def logistic_function(z):
    return 1 / (1 + np.exp(-z))

def predict(x, theta):
    z = np.dot(x, theta)
    y_hat = logistic_function(z)
    return z, y_hat

def compute_loss(y_hat, y):
    m = len(Y)
    loss = sum(-1 * (y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))) / m
    return loss

def compute_gradient(x, y_hat, y):
    gradient = x.dot((y_hat - y))
    return gradient / len(Y)

def update_weight(gradient, theta, lr):
    theta = theta - lr * gradient
    return theta


```

```python
z, y_hat = predict(X, theta)
print(y_hat)
```

```python
loss = compute_loss(y_hat, Y)
print(loss)
```

```python
gradient = compute_gradient(X.T, y_hat, Y)
print(gradient.shape)
print(gradient)
```

```python
theta = update_weight(gradient, theta, 0.1)
print(theta)
```
