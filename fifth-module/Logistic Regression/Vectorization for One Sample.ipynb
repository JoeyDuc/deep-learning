```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
iris = np.genfromtxt('data/iris_1D.csv', delimiter=',', skip_header=1)
print(iris.shape)
X = iris[:, 0]
y = iris[:, 1]
print(f'X = {X}')
print(f'y = {y}')
```

```python
# Visualization
plt.figure(figsize=(6, 4))
plt.scatter(X[y == 0][:], y[y == 0], color = 'b', label = '0')
plt.scatter(X[y == 1][:], y[y == 1], color = 'r', label = '1')
plt.legend()
```

```python
def predict(x, w, b):
    z = x * w + b
    y_hat = 1 / (1 + np.exp(-z))
    return y_hat

def compute_loss(y, y_hat):
    return -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))

def compute_gradient(x, y_hat, y):
    dl_dw = x * (y_hat - y)
    dl_db = y_hat - y
    return dl_dw, dl_db

def update_weight(w, b, dw, db, lr):
    w = w - lr * dw
    b = b - lr * db
    return w, b
```

```python
w, b = -0.1, 0.1
losses = []
preds = []
for _ in range(100):
    # for each sample:
    for i in range(len(y)):
        # a sample
        xi = X[i]
        yi = y[i]
        # compute output
        y_hat = predict(xi, w, b)
        preds.append(y_hat)
        
        # compute loss
        loss = compute_loss(yi, y_hat)
        losses.append(loss)
        # compute gradient
        dl_dw, dl_db = compute_gradient(xi, y_hat, yi)

        # update weight
        w, b = update_weight(w, b, dl_dw, dl_db, 0.1)
        
```

```python
print(f'w = {w}, b = {b}')
```

```python
plt.plot(losses)
plt.show()
```
